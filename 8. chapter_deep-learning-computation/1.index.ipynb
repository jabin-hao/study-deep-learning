# 深度学习计算
:label:`chap_computation`

深度学习的快速发展离不开三个重要因素：海量数据、强大的计算硬件和优秀的软件工具。

深度学习框架（如 PyTorch）提供了高级 API，使开发者能够快速构建神经网络，减少重复代码，同时支持底层自定义。

本章主要介绍以下内容：

1. 模型构建：使用层（Layer）和块（Block）组织神经网络。
2. 参数管理：访问、修改和初始化模型参数。
3. 自定义层与块：根据需求设计新的网络组件。
4. 模型保存与加载：将模型参数保存到磁盘并重新读取。
5. GPU 计算：利用 GPU 加速模型训练和推理。

核心目的： 掌握 PyTorch 的高级功能，从能够使用神经网络，进阶到能够灵活构建、管理和优化神经网络。

这一章不重点介绍新模型，而是为后续学习 CNN、ResNet、Transformer 等复杂网络打下工程基础。
