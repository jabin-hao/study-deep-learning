## 延后初始化（Deferred Initialization）

延后初始化是指：创建神经网络时，暂时不确定输入维度，直到第一次传入数据时，再自动推断维度并初始化参数。

### 1. 为什么需要延后初始化？

定义神经网络时，可能存在以下问题：

- 不知道输入数据的维度。
- 不知道前一层的输出维度。
- 无法提前确定模型参数的数量和形状。

延后初始化可以解决这些问题，使网络的定义和修改更加方便。

### 2. 实现原理

```
创建网络
    ↓
暂不确定部分参数形状
    ↓
第一次传入数据
    ↓
自动推断输入维度
    ↓
初始化对应参数
    ↓
正常进行前向传播
```

### 3. PyTorch 示例

```
# 创建网络，不指定第一层的输入维度net = nn.Sequential(    nn.LazyLinear(256),    nn.ReLU(),    nn.Linear(256, 10))# 第一次前向传播，自动推断输入维度X = torch.rand(2, 20)net(X)# 第一层权重形状为(256, 20)print(net[0].weight.shape)
```

### 4. 核心总结

延后初始化 = 先定义网络，后根据输入数据初始化参数。

在 PyTorch 中，可以通过 `nn.LazyLinear` 实现延后初始化，减少手动指定输入维度的工作。

此时，因为输入维数是未知的，所以网络不可能知道输入层权重的维数。
因此，框架尚未初始化任何参数，我们通过尝试访问以下参数进行确认。


接下来让我们将数据通过网络，最终使框架初始化参数。


一旦我们知道输入维数是20，框架可以通过代入值20来识别第一层权重矩阵的形状。
识别出第一层的形状后，框架处理第二层，依此类推，直到所有形状都已知为止。
注意，在这种情况下，只有第一层需要延迟初始化，但是框架仍是按顺序初始化的。
等到知道了所有的参数形状，框架就可以初始化参数。

## 小结

* 延后初始化使框架能够自动推断参数形状，使修改模型架构变得容易，避免了一些常见的错误。
* 我们可以通过模型传递数据，使框架最终初始化参数。
