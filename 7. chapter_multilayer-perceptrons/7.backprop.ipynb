# 前向传播、反向传播和计算图

## 1. 前向传播（Forward Propagation）

**前向传播是指数据从输入层到输出层，逐层计算预测结果和损失的过程。**

以单隐藏层神经网络为例（忽略偏置）：

**① 隐藏层线性变换**

$$
\mathbf{z}=\mathbf{W}^{(1)}\mathbf{x}
$$

**② 激活函数**

$$
\mathbf{h}=\phi(\mathbf{z})
$$

**③ 输出层计算**

$$
\mathbf{o}=\mathbf{W}^{(2)}\mathbf{h}
$$

**④ 计算损失**

$$
L=l(\mathbf{o},y)
$$

如果使用 $L_2$ 正则化，还需要计算正则化项：

$$
s=\frac{\lambda}{2}\left(\|\mathbf{W}^{(1)}\|_F^2+\|\mathbf{W}^{(2)}\|_F^2\right)
$$

最终目标函数：

$$
J=L+s
$$

其中：

- $\mathbf{x}$：输入数据。
- $\mathbf{W}$：模型权重。
- $\phi$：激活函数。
- $\mathbf{o}$：预测输出。
- $L$：预测损失。
- $s$：正则化惩罚项。
- $J$：需要最小化的目标函数。

## 2. 计算图（Computational Graph）

计算图用于表示神经网络中**变量和运算之间的依赖关系**。

前向传播沿计算图从输入到输出计算；反向传播则沿相反方向计算梯度。

## 3. 反向传播（Backpropagation）

**反向传播是指从输出层到输入层，利用链式法则计算各层参数梯度的过程。**

其目的是确定每个参数应该如何调整，才能减小损失。

链式法则：

$$
\frac{dz}{dx}=\frac{dz}{dy}\cdot\frac{dy}{dx}
$$

对于前面的神经网络：

**① 计算输出层权重梯度**

$$
\frac{\partial J}{\partial \mathbf{W}^{(2)}}
=
\frac{\partial J}{\partial \mathbf{o}}\mathbf{h}^{\top}
+\lambda\mathbf{W}^{(2)}
$$

**② 将梯度传播到隐藏层**

$$
\frac{\partial J}{\partial \mathbf{h}}
=
(\mathbf{W}^{(2)})^\top
\frac{\partial J}{\partial \mathbf{o}}
$$

**③ 经过激活函数反向传播**

$$
\frac{\partial J}{\partial \mathbf{z}}
=
\frac{\partial J}{\partial \mathbf{h}}\odot\phi'(\mathbf{z})
$$

**④ 计算隐藏层权重梯度**

$$
\frac{\partial J}{\partial \mathbf{W}^{(1)}}
=
\frac{\partial J}{\partial \mathbf{z}}\mathbf{x}^{\top}
+\lambda\mathbf{W}^{(1)}
$$

其中，$\odot$ 表示逐元素相乘，$\lambda\mathbf{W}$ 是 $L_2$ 正则化产生的梯度项。

## 4. 神经网络训练流程

1. **前向传播**：计算模型预测值。
2. **计算损失**：衡量预测值与真实值的差距。
3. **反向传播**：计算模型参数的梯度。
4. **参数更新**：通过优化器更新权重。
5. **重复迭代**：不断降低损失，优化模型。

PyTorch 中的典型实现：

```python
optimizer.zero_grad()       # 清空梯度
y_hat = net(X)              # 前向传播
loss = criterion(y_hat, y)  # 计算损失
loss.backward()             # 反向传播
optimizer.step()            # 更新参数
```

## 5. 为什么训练比预测更占显存？

- **训练阶段**：需要保存前向传播的中间结果，用于反向传播计算梯度。
- **预测阶段**：通常只进行前向传播，不需要保存用于反向传播的中间结果。

因此，网络越深、批量越大，训练时通常需要越多显存。

## 6. 总结

- **前向传播**：从输入到输出，计算预测结果和损失。
- **反向传播**：从输出到输入，利用链式法则计算梯度。
- **计算图**：记录变量之间的运算与依赖关系。
- **梯度下降**：根据梯度更新模型参数。
- **训练比预测更占显存**：因为需要保存反向传播所需的中间结果。

**核心流程：前向传播 → 计算损失 → 反向传播 → 更新参数 → 重复训练。**
