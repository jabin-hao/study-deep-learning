# 卷积神经网络
:label:`chap_cnn`

## 卷积神经网络（CNN）

### 1. 为什么需要卷积神经网络？

传统多层感知机（MLP）处理图像时，需要将二维图像展平为一维向量，导致图像的空间结构信息无法得到充分利用。

而图像中相邻像素通常具有较强的关联性，因此需要能够有效利用空间信息的网络。

### 2. 什么是卷积神经网络？

卷积神经网络（Convolutional Neural Network，CNN）是一类**专门用于处理图像等具有空间结构的数据**的神经网络。

与传统全连接网络相比，CNN 具有以下优势：

- 保留空间结构：利用相邻像素之间的关联性提取特征。
- 参数更少：通过局部连接和参数共享减少模型参数。
- 计算高效：卷积运算适合 GPU 并行加速。
- 应用广泛：用于图像分类、目标检测、语义分割等任务，也可应用于音频、文本和时间序列。

### 3. 本章主要内容

| 知识点              | 作用              |
| ---------------- | --------------- |
| 卷积层（Convolution） | 提取图像的局部特征       |
| 填充（Padding）      | 控制特征图的空间尺寸      |
| 步幅（Stride）       | 控制卷积核的移动距离      |
| 池化层（Pooling）     | 汇聚局部信息，降低空间分辨率  |
| 多通道（Channel）     | 处理多通道输入并提取多种特征  |
| LeNet            | 学习经典卷积神经网络的完整实现 |

### 4. 核心总结

CNN 的核心思想：利用图像的局部空间结构，通过卷积操作高效提取特征。

相比将图像展平后输入 MLP，CNN 更适合处理具有空间结构的图像数据，是计算机视觉的重要基础。
